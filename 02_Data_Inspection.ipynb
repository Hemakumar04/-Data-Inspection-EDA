## 9. `info()`

`info()` gives a concise summary of the dataframe: number of entries, column
names, non-null counts (helps spot missing values), data types, and memory
usage.


```python
df.info()
```

**Observation:** Columns like `tenure_months`, `internet_service`,
`payment_method`, and `total_charges` show fewer non-null entries than the

we will handle in a later sprint (data cleaning).


## 10. `describe()`

`describe()` generates summary statistics. By default it covers only
**numerical columns** (count, mean, std, min, quartiles, max).


```python
df.describe()
```

We can pass `include='object'` to get summary statistics for
**categorical (text) columns** instead — count, number of unique values,
the most frequent value (`top`), and its frequency (`freq`).


```python
df.describe(include='object')
```

**Observation:** `describe()` on numerical columns shows a **negative
minimum value for `monthly_charges`**, which is not realistic (charges
cannot be negative) — this is a data-entry error we intentionally kept in
this dataset to demonstrate a real-world data-quality issue.

## 14. Column Classification

Based on the inspection above, we classify every column into a category.

| Column | Type | Notes |
|---|---|---|
| `customer_id` | **Identifier** | Unique (or near-unique) label per row; not used as a model feature |
| `gender` | Categorical | 2 main categories, with some inconsistent spelling variants (data-quality issue) |
| `senior_citizen` | Numerical (Discrete) | Only takes values 0 or 1 (flag) |
| `partner` | Categorical | Yes/No, with some inconsistent spelling variants |
| `dependents` | Categorical | Yes/No |
| `tenure_months` | Numerical (Discrete) | Whole number of months; has some missing values |
| `phone_service` | Categorical | Yes/No |
| `internet_service` | Categorical | DSL / Fiber optic / No; has some missing values |
| `contract` | Categorical | Month-to-month / One year / Two year |
| `paperless_billing` | Categorical | Yes/No |
| `payment_method` | Categorical | 4 payment types; has some missing values |
| `monthly_charges` | Numerical (Continuous) | Contains a data-entry error (negative values) |
| `total_charges` | Numerical (Continuous), stored as **text (object)** | Needs type conversion; has missing values |
| `churn` | Categorical — **Target Variable** | Yes/No — what we want to predict |

**No date/time columns** are present in this dataset (this dataset does not
track signup dates or timestamps — it uses `tenure_months` as a proxy for
customer duration instead).

### Column Groups Summary


```python
identifier_columns = ['customer_id']
target_column = ['churn']
numerical_columns = ['senior_citizen', 'tenure_months', 'monthly_charges', 'total_charges']
categorical_columns = ['gender', 'partner', 'dependents', 'phone_service',
                        'internet_service', 'contract', 'paperless_billing',
                        'payment_method']
datetime_columns = []  
print("Identifier column(s) :", identifier_columns)
print("Target column        :", target_column)
print("Numerical columns    :", numerical_columns)
print("Categorical columns  :", categorical_columns)
print("Date/time columns    :", datetime_columns if datetime_columns else "None present in this dataset")
assert len(identifier_columns) + len(target_column) + len(numerical_columns) + len(categorical_columns) == df.shape[1]
print("\nColumn classification accounts for all", df.shape[1], "columns. ✔")
```

## 15. Data-Quality Issues Identified During Inspection

Summary of the realistic issues found while inspecting this dataset (to be
addressed in a future data-cleaning sprint):

1. **Missing values** in `tenure_months`, `internet_service`,
   `payment_method`, and `total_charges` (seen via `info()` and
   `isnull().sum()`).
2. **Wrong data type** — `total_charges` is stored as `object` (text)
   instead of a numeric type, partly due to stray leading spaces.
3. **Inconsistent categorical labels** — e.g. `gender` contains
   `'Male'`, `'M'`, `'male'` as separate values instead of one consistent
   label (seen via `value_counts()`).
4. **Duplicate rows** — some customer records appear more than once
   (seen via `nunique()` on `customer_id` being less than total row count).
5. **Invalid values** — `monthly_charges` contains a few negative values,
   which is not physically meaningful for a charge amount (seen via
   `describe()`).


```python
print("Duplicate rows in dataset:", df.duplicated().sum())
print("Rows with negative monthly_charges:", (df['monthly_charges'] < 0).sum())
```
