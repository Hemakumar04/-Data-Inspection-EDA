```python
import pandas as pd
import numpy as np
pd.set_option('display.max_columns', None)
df = pd.read_csv('telecom_customer_churn.csv')
df.head()
```

## 2. What is a Dataset?

A **dataset** is a structured collection of data, typically organized in a
table (rows and columns), that is used for analysis or to train a Machine
Learning model.

- Each **row** represents a single observation / record (here, one customer).
- Each **column** represents a **feature** (a piece of information about that
  observation).

Our dataset is a table of telecom customers, where each row describes one
customer and whether they churned.


```python
print("Dataset shape (rows, columns):", df.shape)
print("\nThis dataset has", df.shape[0], "records (customers) and", df.shape[1], "columns (features).")
```

## 3. What is a Feature?

A **feature** is an individual measurable property or characteristic of the
data being observed. In a tabular dataset, **every column (except the
target/identifier) is a feature**.

Example features in our dataset: `gender`, `tenure_months`, `contract`,
`monthly_charges`, etc.


```python
features = [col for col in df.columns if col not in ['customer_id', 'churn']]
print("Number of features:", len(features))
print("\nFeatures:\n", features)
```

## 4. What is a Target?

The **target** (also called the **label** or **outcome variable**) is the
column we want to **predict**. It is what the Machine Learning model learns
to output based on the features.

In our dataset, the target is **`churn`** — whether a customer will leave
(`Yes`) or stay (`No`) with the telecom company. This makes it a **binary
classification problem**.


```python
print("Target variable: churn")
df['churn'].value_counts()
```

## 5. Independent Variables vs Dependent Variables

- **Independent Variables (Features / Inputs):** These are the columns used
  to make predictions. Their values do not depend on the outcome — e.g.
  `tenure_months`, `contract`, `monthly_charges`.
- **Dependent Variable (Target / Output):** This is the column whose value we
  are trying to predict, and it **depends** on the independent variables. In
  our case, this is `churn`.

In short: **Independent variables → (model) → Dependent variable**


```python
independent_variables = [col for col in df.columns if col not in ['customer_id', 'churn']]
dependent_variable = 'churn'
print("Independent variables (X):", independent_variables)
print("\nDependent variable (y):", dependent_variable)
```

## 6. Numerical Features

**Numerical features** contain quantitative, measurable values (numbers) on
which mathematical operations make sense.

In our dataset: `senior_citizen` (0/1 flag), `tenure_months`,
`monthly_charges`, and `total_charges` (once cleaned from text) are numerical.


```python
numerical_features = ['senior_citizen', 'tenure_months', 'monthly_charges', 'total_charges']
df[numerical_features].dtypes
```

## 7. Categorical Features

**Categorical features** contain a limited, fixed set of distinct
categories/labels rather than numbers. They describe a quality or grouping.

In our dataset: `gender`, `partner`, `dependents`, `phone_service`,
`internet_service`, `contract`, `paperless_billing`, `payment_method`, and
the target `churn` are all categorical.


```python
categorical_features = ['gender', 'partner', 'dependents', 'phone_service',
                        'internet_service', 'contract', 'paperless_billing',
                        'payment_method', 'churn']
for col in categorical_features:
    print(f"{col}: {df[col].unique()}")
```

## 8. Discrete vs Continuous Features

Numerical features can be further split into:

- **Discrete features** — take only specific, countable values (often whole
  numbers), e.g. `tenure_months` (0, 1, 2, ... 72 — a count of months) and
  `senior_citizen` (only 0 or 1).
- **Continuous features** — can take *any* value within a range, including
  decimals, e.g. `monthly_charges` and `total_charges` (amounts of money).


```python
discrete_features = ['tenure_months', 'senior_citizen']
continuous_features = ['monthly_charges', 'total_charges']
print("Discrete features:", discrete_features)
print("Continuous features:", continuous_features)
print("\nExample - unique values of a discrete feature (tenure_months):")
print(sorted(df['tenure_months'].dropna().unique())[:15], "...")
```

## 9. Identifier Columns

An **identifier column** uniquely identifies each row/record but carries **no
predictive/statistical meaning** for modeling. It should typically be
**excluded from model training**.

In our dataset, `customer_id` is the identifier column — every value is
(mostly) unique and just labels the row.


```python
print("Is customer_id unique for every row?")
print("Unique values:", df['customer_id'].nunique(), "out of", len(df), "rows")
print("(Fewer unique values than rows indicates duplicate customer records - a data quality issue we'll handle later.)")
```

## 10. Target Variable (Recap)

As established above, **`churn`** is the target variable for this dataset —
a categorical column with two classes: `Yes` and `No`. Predicting this
column is the core business goal (identify customers likely to leave, so the
company can act to retain them).


```python
df['churn'].value_counts(normalize=True) * 100  # class distribution in %
```

## 11. Metadata

**Metadata** is "data about the data" — information describing the dataset
itself rather than its actual values. This includes column names, data
types, number of rows/columns, memory usage, and general structure.


```python
print("=== Dataset Metadata ===")
print("Number of rows      :", df.shape[0])
print("Number of columns   :", df.shape[1])
print("Column names        :", list(df.columns))
print("\nData types of columns:")
print(df.dtypes)
```

## 12. Dataset Dimensions

The **dimensions** of a dataset refer to its shape — the number of rows
(records) by the number of columns (features), typically written as
`(rows, columns)`.


```python
print("Dataset dimensions:", df.shape)
print(f"This dataset has {df.shape[0]} rows and {df.shape[1]} columns.")
```
